## Del 1: Trafikkmodell: tetthet, hastighet og fluks

Vi skal modellere trafikkflyten langs en veistrekning. Vi innfører flere forenklinger:

- det er ingen kryss 
- ingen forbikjøringer
- kun en fil
- ingen kjører av eller på veien
- alle bilene er like lange
- trafikken flyter en vei
- alle sjåførene har samme kjøreadferd, som inkluderer at alle holder seg til fartsgrensen.
- m. m.

Merk at det er fullt mulig å modellere trafikken med et system av ordinære differensiallikninger hvor vi 
regner på posisjon, hastighet og akselerasjon til hver enkelt bil. Det ville vært en såkalt *mikroskopisk* 
modell.

Det er derimot *ikke* hva vi skal gjøre i denne oppgaven. Vi skal se på en *makroskopisk* modell.

Vi modellerer trafikken som den ser ut observert fra noen som står på siden av veien. Det observeres alt for 
mange biler til at vi kan se på hver enkelt bil. Istedet beskriver vi trafikken ved hjelp av funksjonen 
$u(x,t)$ som gir tettheten av biler i posisjon $x$ langs veien ved tidspunkt $t$. 

Vi tenker oss at $u(x,t)$ for et tidspunkt $t$ er gitt ved antall biler i intervallet $[x-R,x+R]$ delt på 
lengden av intervallet $2R$, hvor lengden på intervallet $R$ er et passende tall.

Måleenhet for $u(x,t)$ er derfor antall biler per meter. Vi har alltid $u(x,t)\geq 0.$

Hastigheten $v(x,t)$ til bilene ved posisjon $x$ og tidspunkt $t$ er også viktig. 

#### Vår viktigste antakelse er at hastigheten $v(x,t)$ kun er avhengig av tettheten $u(x,t)$. Dvs.  hastigheten $v=v(u)$ er en funksjon av $u$.

For hastigheten $v$ så antar vi i tillegg at:

1. $v\geq 0$  
2. $v\leq v_{max}$, hvor $v_{max}$ er maksfart, dvs. fartsgrensen på veien.
3. $\frac{dv}{du}\leq 0$.
4. $v(0) = v_{max}$ når tettheten $u=0$.
5. $v(u_{max}) = 0$ når tettheten $u=u_{max}$ er størst mulig. 

Det er mange mulige valg av funksjoner $v=v(u)$. I praksis er dette noe vi må måle på den aktuelle 
veistrekningen. Det blir derimot for komplisert og tidkrevende for oss. Istedet prøver vi med noen funksjoner 
som kan passe.

Den enkleste mulige funksjonen er den lineære funksjonen 
$$
  v=v_{max}(1-\frac{u}{u_{max}}).
$$

En annen mulig funksjon er den kvadratiske funksjonen
$$
  v=v_{max}(1-(\frac{u}{u_{max}})^2).
$$

Dette er på ingen måte de eneste mulighetene for $v(u)$. Vi gir flere eksempler lenger nede i oppgaven, og vi vil også at dere skal prøve å lage noen slike funksjoner.

Vi skal også se på strømmen eller *fluksen* av biler gjennom et punkt $x$ langs veien ved tiden $t$. 

Vi definerer fluksen av biler, $J(x,t)$, til å være antall biler som passerer et punkt $x$ per sekund. Altså har $J$ måleenhet antall biler per sekund. Vi har sammenhengen
$$
  J(x,t)=J(u(x,t))=u(x,t)v(u(x,t)).
$$
og alltid at $J(x,t)\geq 0$.

Antall biler som passerer punktet $x$ på veien i tidsintervallet $[t,t+\Delta t]$ er altså (tilnærmet lik)
$$
  J(x,t)\Delta t
$$

Videre er antall biler som befinner seg i intervallet $[x,x+\Delta x ]$ ved tiden $t$ gitt ved (tilnærmet) 
$$
  u(x,t)\Delta x
$$

Nå skal vi sette opp et regnskap for bilene som er helt analogt med det de fleste av oss har et forhold til. Vi vet at penger inn på konto minus penger ut av konto over et tidsrom er lik netto endring av saldoen.

Tilsvarende, er antall biler i intervallet $[x,x+\Delta x]$ ved tiden $t+\Delta t$ minus antall biler i $[x,x+\Delta x]$ ved tiden $t$ lik netto endring i antall av biler i $[x,x+\Delta x]$ i tidsrommet $t$ til $t+\Delta t$.

Analogt med bankkonto må dette være likt antall biler som kjører inn i intervallet $[x,x+\Delta x]$ minus antall biler som kjører ut av $[x,x+\Delta x]$ i tidsrommet $t$ til $t+\Delta t$.

Uttrykker vi regnskapet matematisk, så fører det til ligningen
$$
    \frac{u(x,t+\Delta t)-u(x,t)}{\Delta t}=-\frac{J(x+\Delta x, t)-J(x,t)}{\Delta x}.
$$
Som vanlig i denne type modellering ser vi på ligning over når $\Delta x\rightarrow 0$ og $\Delta t\rightarrow 0$. 

Ligningen over blir da den partielle differensiallikningen
$$
    \frac{\partial u(x,t)}{\partial t}=-\frac{\partial J(x,t)}{\partial x}
$$
som gjelder for verdier $x\in [a,b]$ og $t\geq 0$.  

Vi har at også $J$ er en funksjon av $u$, gitt ved $J(u)=u\cdot v(u)$, slik at likningen til slutt kan skrives som 
$$
  u_t + (u\cdot v(u))_x = 0
$$

I tillegg så vil en fullstendig problembeskrivelse ha mer informasjon. For eksempel randbetingelser som $u(a,t)=u_a(t)$ som beskriver hvor mange biler som kommer inn i veistrekningen, og initialbetingelser som $u(x,0)=u_0(x)$ som beksriver tettheten av bilder når tiden $t=0$ starter.
