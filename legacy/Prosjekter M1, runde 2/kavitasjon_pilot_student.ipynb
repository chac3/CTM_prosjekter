# Del A: Trykk og volumstrøm i et rørsystem

## A.1 Tre begreper

### Volumstrøm

Volumstrømmen $Q$ forteller hvor stort væskevolum som passerer et tverrsnitt per tidsenhet. Den kan for eksempel måles i liter per sekund.

### Trykk

Trykk kan sammenlignes med elektrisk potensial. Væske strømmer vanligvis fra et område med høyere trykk mot et område med lavere trykk.

### Damptrykk

Ved en gitt temperatur har væsken et damptrykk $p_v$. Dersom det lokale **absolutte trykket** blir lavere enn damptrykket, kan det dannes dampbobler.

I denne delen bruker vi en enkel lineær modell for trykktap i et rør:

$$
\Delta p=KQ.
$$

Her er $K$ en hydraulisk motstand. Denne lineære modellen er en forenkling. I mange virkelige rørsystemer er trykktapet ikke-lineært.

- $Q_1$ være volumstrømmen i rør 1,

Dersom radiusen bare avviker litt fra likevekten, skriver vi

$$R(t)=R_0+y(t),$$

# Del C: Ikke-lineær boblevekst og kollaps

## C.1 Rayleigh–Plesset-ligningen

En vanlig modell for en sfærisk boble i en stor væskemengde er

$$
R\ddot R+\frac{3}{2}\dot R^2
=
\frac{1}{\rho}
\left[
 p_B(R)-p_\infty(t)
 -\frac{2\sigma}{R}
 -\frac{4\mu\dot R}{R}
\right].
$$

Symbolene betyr:

- $R(t)$: bobleradius,
- $\rho$: væsketetthet,
- $\mu$: dynamisk viskositet,
- $\sigma$: overflatespenning,
- $p_B(R)$: trykk inne i boblen,
- $p_\infty(t)$: trykk i væsken et stykke fra boblen.

Leddet med $\dot R^2$ gjør ligningen ikke-lineær. Det samme gjør faktorene $1/R$.

Vi modellerer trykket inne i boblen som summen av damptrykk og komprimert gass:

$$
p_B(R)=p_v+p_{g0}\left(\frac{R_0}{R}\right)^{3\gamma}.
$$

Her er $\gamma$ en gitt konstant. Gasstrykket ved likevekt velges slik at $R=R_0$ faktisk er en likevekt:

$$p_{g0}=p_{\infty,0}+\frac{2\sigma}{R_0}-p_v.$$

## C.2 Et midlertidig trykkfall

I stedet for å modellere hele strømningen gjennom en ventil eller pumpe, gir vi væsketrykket direkte:

$$
p_\infty(t)
=
p_{\infty,0}
-\Delta p\exp\left[-\frac{(t-t_c)^2}{2\tau^2}\right].
$$

Dette representerer at bobla passerer et område med lavt trykk:

- før passasjen er trykket omtrent $p_{\infty,0}$,
- nær $t=t_c$ er trykket lavest,
- etter passasjen stiger trykket igjen.

Dette trykkforløpet er en pedagogisk inngangsfunksjon, ikke resultatet av en full strømningsberegning.

````python
rho = 998.0          # kg/m^3
mu = 1.0e-3          # Pa s
sigma = 0.072        # N/m
p_v = 2.34e3         # Pa
p_inf0 = 101.3e3     # Pa
R0 = 50.0e-6         # m
gamma = 1.4

Delta_p = 35.0e3     # Pa
t_c = 120.0e-6       # s
tau = 35.0e-6        # s

p_g0 = p_inf0 + 2*sigma/R0 - p_v


def p_inf(t):
    return p_inf0 - Delta_p*np.exp(-0.5*((t - t_c)/tau)**2)


def p_boble(R):
    return p_v + p_g0*(R0/R)**(3*gamma)


t_plott = np.linspace(0.0, 300e-6, 1000)
plt.plot(t_plott*1e6, p_inf(t_plott)/1e3)
plt.xlabel("Tid i mikrosekunder")
plt.ylabel("Væsketrykk i kPa")
plt.grid()
plt.show()
````

## Oppgave D1: Forutsetninger

Diskuter minst fire av disse forutsetningene:

- Bobla er kuleformet.
- Væsken er uendelig stor sammenlignet med bobla.
- Væsketettheten er konstant.
- Trykket inne i bobla er uniformt.
- Bobla er langt fra vegger og andre bobler.
- Trykkforløpet er gitt og påvirkes ikke av bobla.
- Modellen tar ikke fullt hensyn til væskekompressibilitet ved voldsom kollaps.
- Modellen beskriver en eksisterende boblekjerne, ikke hvordan kjernen først oppstår.

## Oppgave D2: Et mulig venturirør

I en fordypningsoppgave kan trykkforløpet beregnes fra et rør med varierende tverrsnitt. For en inkompressibel væske er

$$Q=A(s)u(s),$$

slik at liten flate $A$ gir stor hastighet $u$. En forenklet Bernoulli-modell kobler deretter stor hastighet til lavt trykk. Denne utvidelsen krever mer fluidmekanikk og er derfor ikke nødvendig i hovedprosjektet.
