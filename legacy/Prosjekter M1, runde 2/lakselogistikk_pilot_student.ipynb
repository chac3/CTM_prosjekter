$$\dot{\mathbf u}=D(\mathbf u)(\mathbf b-A\mathbf u),\qquad \mathbf u=(x,y,z)^T,$$

$$A=\begin{pmatrix}1&0.30&0.20\\-0.40&1&0.10\\0&0.20&1\end{pmatrix},\qquad
\mathbf b=\begin{pmatrix}1.28\\0.25\\0.62\end{pmatrix},$$

$$D(\mathbf u)=\operatorname{diag}(x,0.7y,0.5z).$$

$$\dot x=x(1.28-x-0.30y-0.20z),$$
$$\dot y=0.7y(0.25+0.40x-y-0.10z),$$
