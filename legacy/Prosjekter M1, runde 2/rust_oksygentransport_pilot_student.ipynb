Vi begynner med en redusert modell. La

- $C(t)$ være en skalert oksygenkonsentrasjon ved ståloverflaten,
- $C_{ute}$ være oksygenkonsentrasjonen i omgivelsene,
- $R(t)$ være skalert rusttykkelse,
- $\alpha>0$ beskrive tilførsel av oksygen fra omgivelsene,
- $k>0$ beskrive forbruk av oksygen i korrosjonsreaksjonen,
- $\kappa>0$ koble oksygenforbruk til rustvekst.

Vi foreslår modellen

$$
\begin{aligned}
\dot C &= \alpha(C_{ute}-C)-kC,\\
\dot R &= \kappa kC.
\end{aligned}
$$

Uttrykket $\alpha(C_{ute}-C)$ betyr at oksygen transporteres inn når konsentrasjonen utenfor er større enn konsentrasjonen ved stålet. Uttrykket $kC$ betyr at mer tilgjengelig oksygen gir større oksygenforbruk per tidsenhet. Dette er en enkel lineær reaksjonslov, ikke en full elektrokjemisk korrosjonsmodell.
