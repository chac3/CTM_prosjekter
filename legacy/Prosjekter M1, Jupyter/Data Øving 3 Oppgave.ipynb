## Utregning av egenverdier i numpy

En kollonnevektor $v$ er en egenvektor med tilhørende egenverdi $\lambda$ hvis $M\cdot v = \lambda \cdot v$. 
Vi sier at $v$ er en høyre egenvektor, da $v$ står til høyre for $M$.

En radvektorer $v$ er en (venstre) egenvektor hvis $v \cdot N = \lambda \cdot v$.

Vi kan bruke np.linalg.eig for å regne ut egenvektorer og egenverdier i numpy.

    import numpy as np   
    A = np.array([[1.2, 3.2], [4.1, 5.5]])
    
    # Høyre egenvektorer, dvs. egenvektorer x hvor Ax = l*x, hvor l er tilhørende egenverdi
    eigenval, eigenvec = np.linalg.eig(A)   
    
    for i in range(len(eigenval)):
        print("Egenvektor",eigenvec[:,i],"med tilhørende egenverdi",eigenval[i])
