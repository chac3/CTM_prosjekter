## Jukselapp

Man kan lett løse et system som har et entydig svar i Python:

```python
import numpy as np

A = np.array([[1,2],[3,4]])
b = np.array([5,6])

u = np.linalg.solve(A,b)
print(u)
```

## Homogene ligninger

Ellers hvis man vil finne nullrommet til et system er det ikke en like enkel sak. Den beste metoden er å bruke SVD (singular value decomposition), som ikke er pensum. Prat med din numerikklærer for å finne ut hvorfor vi liker det bedre enn redusert trappeform. Vi finner SVD under:

```python
C = np.array([[1,2,3],[4,5,6],[5,7,9]])

U, S, V = np.linalg.svd(C)
print(S)
```

Hvis et av tallene over er lik (eller nesten lik) null, er nullrommet ikke tomt. Her ser vi at den 3. tall i S er tilnærmet null. Da er den 3. raden i V i nullrommet. Siden python teller fra 0 finner vi det ved å sette i=3-1=2 under:

```python
i = 2
print(V[i,:]/V[i,0])
```
