## Moment og skjærkraft (fritt bjelke)

Vi har $M' = V$, hvor $V$ er skjærkraft og $M$ bøyemoment i en bjelke.

Det er en differensialligning som kan løses med integrasjon, om vi er oppgitt $V=V(x)$. For enkle eksempler kan vi gjøre det for hånd, ellers må vi kjøre numerisk

## Den elastiske linjen

Vi skal snart se at moment er relatert til forskyvningen $w(x)$ gjennom ligningen 

$$M = -EIw''(x).$$ 

Det kalles for ligningen for den elastiske linjen. Det kan kombineres med et kjent $M(x)$. Alternativt kan vi kombiner ligningene og få 

$$w''''(x) = \frac{p}{EI},$$

hvor $p=-V'$ er den fordelte lasten. Den kalles for bjelkens ligning.

## Knekning

En klassisk problemstilling gjelder knekning av en søyle. Der har vi $M=Pw$, hvor $P$ er den normale kraften på søyla. Da får vi en ligning

$$
w'' + \frac{P}{EI} w = 0,
$$

Med ranbetingelsene...

Ellers kan vi ta andre bjelker i samme situasjon. For eksempel en fritt bjelke gir

$$
w'''' + \frac{P}{EI}w'' = 0,
$$

Med tilsvarende randbetingelser
