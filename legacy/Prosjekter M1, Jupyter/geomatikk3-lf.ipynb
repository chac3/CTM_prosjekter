## Diagonalisering og ellipsoiden

La oss ta et koordinatsystem $(\vec{x},\vec{y},\vec{z})$ hvor $\vec{y}$ peker mot sola, $\vec{x}$ ligger i planet av jordens bane rundt sola, og peker i retning av jordens bevegelse, og $\vec{z} = \vec{x}\times\vec{y}$.

Ellipsoiden defineres av løsninger til $\vec{r}^T A \vec{r} = 1$, hvor 

$$
A = \begin{pmatrix}
1.456\cdot 10^{-7} & 1.122\cdot 10^{-8} & -3.648\cdot 10^{-8} \\
1.122\cdot 10^{-8} & 1.456\cdot 10^{-7} & 3.648\cdot 10^{-8} \\
-3.648\cdot 10^{-8} & 3.648\cdot 10^{-8} & 3.814\cdot 10^{-8}
\end{pmatrix}.
$$

Ellipsoide? Ligning av en ellipsoide er

$$
\frac{x^2}{a^2} + \frac{y^2}{b^2} + \frac{z^2}{c^2} = 1,
$$

hvor $a, b, c$ er lengde av "semi-aksene", altså avstand fra origo til overflaten langs henholdsvis $x-, y-$ og $z-$aksene.

## Treghetsmoment og "Chandler wobble"

Eulers ligning forklarer rotasjon til et stivt legeme. Den sier at 

\begin{align*}
I_1 \dot{\omega}_1 & = \omega_2 \omega_3 (I_1 - I_3) \\
I_2 \dot{\omega}_2 & = -\omega_1 \omega_3 (I_1 - I_3) \\
I_3 \dot{\omega}_3 & = \omega_1 \omega_2 (I_2 - I_3)
\end{align*}

Her er $\vec{\omega}$ vinkelhastighet: det betyr at legemet roterer rundt en akse som peker i retning $\vec{\omega}$ med hastighet $|\vec{\omega}|$.

Treghetsmoment til en solid ellipsoide (i basisen som diagonaliserer ellipsoiden) med masse $m$ er

$$
I_1 = \frac{m}{5}(b^2 + c^2), \quad
I_2 = \frac{m}{5}(a^2 + c^2), \quad
I_3 = \frac{m}{5}(a^2 + b^2).
$$

I dette tilfellet, du burde finne at den tredje ligningen blir $\dot{\omega}_3 = 0$, altså at $\omega_3$ er konstant. De andre to kan kombineres: la $\Omega = \omega_3(I_1 - I_3)/I_1$, da får vi

$$
\begin{pmatrix}
\dot{\omega}_1 \\
\dot{\omega}_2
\end{pmatrix}
=
\begin{pmatrix}
0 & \Omega \\
-\Omega & 0
\end{pmatrix}
\begin{pmatrix}
\omega_1 \\
\omega_2
\end{pmatrix}
$$

Hvis $\omega_1(0)=\omega_2(0)=0$, vil de forblir null for alle tid. Da får vi ren rotasjon rundt $z$-aksen, med vinkelhastighet $\omega_3$. Det er sikkert hvordan vi tenker på jordens rotasjon. Poenget er at det vil være tendenser til små endringer i jordens rotasjonsakse, siden $\omega_1, \omega_2$ i praksis er liten men ikke helt lik null.
